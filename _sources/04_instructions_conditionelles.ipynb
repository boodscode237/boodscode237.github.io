La programmation implique souvent d'examiner un ensemble de conditions et de décider quelle action entreprendre en fonction de ces conditions. L'instruction `if` de Python vous permet d'examiner l'état actuel d'un programme et de répondre de manière appropriée à cet état.

Dans ce chapitre, vous apprendrez à écrire des tests conditionnels, qui vous permettent de vérifier n'importe quelle condition d'intérêt. Vous apprendrez à écrire des instructions `if` simples, et vous apprendrez comment créer une série plus complexe d'instructions `if` pour identifier quand les conditions exactes que vous souhaitez sont présentes. Vous appliquerez ensuite ce concept aux listes, ce qui vous permettra d'écrire une boucle `for` qui traite la plupart des éléments d'une liste d'une manière, mais gère certains éléments avec des valeurs spécifiques d'une manière différente.

### Un Exemple Simple

L'exemple court suivant montre comment les tests `if` vous permettent de répondre correctement à des situations spéciales. Imaginez que vous ayez une liste de marques de voitures et que vous vouliez imprimer le nom de chaque voiture. Les noms de voitures sont des noms propres, donc les noms de la plupart des voitures doivent être imprimés avec la première lettre en majuscule (casse titre). Cependant, la valeur 'toyota', très courante au Cameroun, devrait être imprimée en majuscules. Le code suivant parcourt une liste de noms de voitures et recherche la valeur 'toyota'. Chaque fois que la valeur est 'toyota', elle est imprimée en majuscules au lieu de la casse titre:

````python
voitures = ['audi', 'toyota', 'mercedes', 'hyundai'] # Liste de voitures, certaines communes au Cameroun
for voiture in voitures:
  if voiture == 'toyota': # Vérifie si la voiture est 'toyota'
    print(voiture.upper()) # Imprime en MAJUSCULES
  else:
    print(voiture.title()) # Imprime avec la première lettre en majuscule
````

La boucle dans cet exemple vérifie d'abord si la valeur actuelle de `voiture` est 'toyota'. Si c'est le cas, la valeur est imprimée en majuscules. Si la valeur de `voiture` est autre que 'toyota', elle est imprimée en casse titre:

```
Audi
TOYOTA
Mercedes
Hyundai
```

Cet exemple combine un certain nombre de concepts que vous apprendrez dans ce chapitre. Commençons par examiner les types de tests que vous pouvez utiliser pour examiner les conditions de votre programme.

### Tests Conditionnels

Au cœur de chaque instruction `if` se trouve une expression qui peut être évaluée comme `True` (Vrai) ou `False` (Faux) et est appelée un test conditionnel. Python utilise les valeurs `True` et `False` pour décider si le code dans une instruction `if` doit être exécuté. Si un test conditionnel est évalué à `True`, Python exécute le code suivant l'instruction `if`. Si le test est évalué à `False`, Python ignore le code suivant l'instruction `if`.

#### Vérification de l'Égalité

La plupart des tests conditionnels comparent la valeur actuelle d'une variable à une valeur spécifique d'intérêt. Le test conditionnel le plus simple vérifie si la valeur d'une variable est égale à la valeur d'intérêt:

````python
voiture = 'toyota'
print(voiture == 'toyota') # Est-ce que voiture est égale à 'toyota' ?
# True
````

La première ligne définit la valeur de `voiture` à 'toyota' en utilisant un seul signe égal. La deuxième ligne vérifie si la valeur de `voiture` est 'toyota' en utilisant un double signe égal (`==`). Cet opérateur d'égalité renvoie `True` si les valeurs à gauche et à droite de l'opérateur correspondent, et `False` si elles ne correspondent pas. Les valeurs dans cet exemple correspondent, donc Python renvoie `True`.

Lorsque la valeur de `voiture` est autre que 'toyota', ce test renvoie `False`:

````python
voiture = 'mitsubishi'
print(voiture == 'toyota') # Est-ce que voiture est égale à 'toyota' ?
# False
````

Un seul signe égal (`=`) est une instruction d'affectation (par exemple, "Définir la valeur de `voiture` à 'mitsubishi'"). Un double signe égal (`==`) pose une question ("La valeur de `voiture` est-elle égale à 'toyota' ?").

#### Ignorer la Casse lors de la Vérification de l'Égalité

Le test d'égalité est sensible à la casse en Python. Par exemple, deux valeurs avec une capitalisation différente ne sont pas considérées comme égales :

````python
voiture = 'Toyota'
print(voiture == 'toyota')
# False
````

Si la casse importe, ce comportement est avantageux. Mais si la casse n'importe pas et que vous voulez simplement tester la valeur d'une variable, vous pouvez convertir la valeur de la variable en minuscules avant de faire la comparaison:

````python
voiture = 'Toyota'
print(voiture.lower() == 'toyota') # Convertit 'Toyota' en 'toyota' avant de comparer
# True
````

Ce test renverrait `True` quelle que soit la façon dont la valeur 'Toyota' est formatée, car le test est désormais insensible à la casse. La méthode `lower()` ne modifie pas la valeur initialement stockée dans `voiture`:

````python
voiture = 'Toyota'
print(voiture.lower() == 'toyota')
# True
print(voiture) # La variable originale n'est pas modifiée
# Toyota
````

La méthode `lower()` est particulièrement utile pour stocker des données. Souvent, vous ne voudrez pas faire confiance à la capitalisation fournie par vos utilisateurs, vous convertirez donc les chaînes en minuscules avant de les stocker.

#### Vérification de l'Inégalité

Lorsque vous voulez déterminer si deux valeurs ne sont *pas* égales, vous pouvez combiner un point d'exclamation et un signe égal (`!=`). Le point d'exclamation représente la négation (`not`).

Utilisons une instruction `if` pour examiner comment utiliser l'opérateur d'inégalité. Stockons un plat préféré dans une variable, puis imprimons un message si la personne n'a pas choisi 'fufu' (un plat populaire en Afrique de l'Ouest et Centrale):

````python
plat_prefere = 'ndole' # Un plat camerounais populaire
if plat_prefere != 'fufu': # Vérifie si le plat préféré N'EST PAS 'fufu'
  print("Tiens bon le fufu !") # Message si ce n'est pas fufu
````

La ligne avec `if` compare la valeur de `plat_prefere` à la valeur 'fufu'. Si ces deux valeurs ne correspondent pas, Python renvoie `True` et exécute le code indenté suivant l'instruction `if`. Si les deux valeurs correspondent, Python renvoie `False` et n'exécute pas le code suivant l'instruction `if`.

Comme la valeur de `plat_prefere` n'est pas 'fufu', la fonction `print()` est exécutée:

```
Tiens bon le fufu !
```

#### Comparaisons Numériques

Tester des valeurs numériques est assez simple. Par exemple, le code suivant vérifie si une personne a 18 ans (âge de la majorité dans de nombreux pays africains) :

````python
age = 18
print(age == 18)
# True
````

Vous pouvez également tester si deux nombres ne sont pas égaux. Par exemple, le code suivant imprime un message si la réponse donnée n'est pas correcte :

````python
reponse_attendue = 42
reponse_utilisateur = 17
if reponse_utilisateur != reponse_attendue: # Si la réponse de l'utilisateur n'est pas 42
  print("Ce n'est pas la bonne réponse. Veuillez réessayer !")
````

Le test conditionnel réussit, car la valeur de `reponse_utilisateur` (17) n'est pas égale à 42. Comme le test réussit, le bloc de code indenté est exécuté:

```
Ce n'est pas la bonne réponse. Veuillez réessayer !
```

Vous pouvez inclure diverses comparaisons mathématiques dans vos instructions conditionnelles, telles que inférieur à (`<`), inférieur ou égal à (`<=`), supérieur à (`>`), et supérieur ou égal à (`>=`):

````python
prix_plantain = 750 # Prix en FCFA
print(prix_plantain < 1000) # Est-ce que le prix est inférieur à 1000 FCFA ?
# True
print(prix_plantain <= 750) # Est-ce que le prix est inférieur ou égal à 750 FCFA ?
# True
print(prix_plantain > 1000) # Est-ce que le prix est supérieur à 1000 FCFA ?
# False
print(prix_plantain >= 1000) # Est-ce que le prix est supérieur ou égal à 1000 FCFA ?
# False
````

#### Vérification de Conditions Multiples

Vous voudrez peut-être vérifier plusieurs conditions en même temps. Par exemple, vous pourriez avoir besoin que deux conditions soient `True` pour effectuer une action, ou être satisfait si une seule condition est `True`. Les mots-clés `and` (et) et `or` (ou) peuvent vous aider dans ces situations.

**Utilisation de `and` pour Vérifier Plusieurs Conditions**

Pour vérifier si deux conditions sont toutes les deux `True` simultanément, utilisez le mot-clé `and` pour combiner les deux tests conditionnels. Si chaque test réussit, l'expression globale est évaluée à `True`. Si l'un ou l'autre des tests échoue, ou si les deux échouent, l'expression est évaluée à `False`.

Par exemple, vous pouvez vérifier si une personne a l'âge légal pour conduire (18 ans au Cameroun) et possède un permis:

````python
age_conducteur = 22
possede_permis = True
print(age_conducteur >= 18 and possede_permis == True)
# True

age_conducteur = 17
print(age_conducteur >= 18 and possede_permis == True)
# False
````

**Utilisation de `or` pour Vérifier Plusieurs Conditions**

Le mot-clé `or` vous permet également de vérifier plusieurs conditions, mais il réussit si l'un ou l'autre, ou les deux, des tests individuels réussissent. Une expression `or` échoue uniquement lorsque les deux tests individuels échouent.

Vérifions si un étudiant a payé ses frais d'inscription ou a obtenu une bourse :

````python
a_paye_inscription = True
a_eu_bourse = False
print(a_paye_inscription == True or a_eu_bourse == True)
# True

a_paye_inscription = False
print(a_paye_inscription == True or a_eu_bourse == True)
# False
````

#### Vérifier si une Valeur est dans une Liste

Il est parfois important de vérifier si une liste contient une certaine valeur avant d'entreprendre une action. Par exemple, vous pourriez vouloir vérifier si un ingrédient est déjà dans la liste des ingrédients d'un plat populaire comme le Ndolé.

Pour savoir si une valeur particulière est déjà dans une liste, utilisez le mot-clé `in`. Faisons une liste d'ingrédients pour le Ndolé et vérifions si certains ingrédients sont dans la liste:

````python
ingredients_ndole = ['feuilles_ameres', 'arachides', 'crevettes', 'huile_rouge']
print('arachides' in ingredients_ndole)
# True
print('viande_de_boeuf' in ingredients_ndole)
# False
````

Les lignes avec `in` indiquent à Python de vérifier l'existence de 'arachides' et 'viande_de_boeuf' dans la liste `ingredients_ndole`.

#### Vérifier si une Valeur n'est pas dans une Liste

Parfois, il est important de savoir si une valeur *n'apparaît pas* dans une liste. Vous pouvez utiliser le mot-clé `not in` dans cette situation. Par exemple, considérez une liste de joueurs suspendus d'une équipe de football. Vous pouvez vérifier si un joueur n'a pas été suspendu avant de l'autoriser à jouer :

````python
joueurs_suspendus = ['etoo', 'mboma', 'kameni']
joueur = 'song'
if joueur not in joueurs_suspendus: # Si 'song' n'est PAS dans la liste des suspendus
  print(f"{joueur.title()}, tu peux jouer le match.")
````

La ligne avec `if` se lit assez clairement. Si la valeur de `joueur` n'est pas dans la liste `joueurs_suspendus`, Python renvoie `True` et exécute la ligne indentée.

Le joueur 'song' n'est pas dans la liste `joueurs_suspendus`, il voit donc un message l'invitant à jouer:

```
Song, tu peux jouer le match.
```

### Instructions `if`

Maintenant que vous comprenez les tests conditionnels, vous pouvez commencer à écrire des instructions `if`. Il existe plusieurs types d'instructions `if`, et votre choix dépend du nombre de conditions que vous devez tester.

#### Instructions `if` Simples

Le type le plus simple d'instruction `if` a un test et une action:

````python
if test_conditionnel:
  faire_quelque_chose
````

Si le test conditionnel est évalué à `True`, Python exécute le code suivant l'instruction `if`. Si le test est évalué à `False`, Python ignore le code suivant l'instruction `if`.

Testons si une personne a l'âge légal pour voter (18 ans au Cameroun):

````python
age = 19
if age >= 18: # Vérifie si l'âge est supérieur ou égal à 18
  print("Vous êtes assez âgé pour voter !")
````

Python vérifie si la valeur de `age` est supérieure ou égale à 18. C'est le cas, donc Python exécute l'appel `print()` indenté:

```
Vous êtes assez âgé pour voter !
```

Vous pouvez avoir autant de lignes de code que vous le souhaitez dans le bloc suivant l'instruction `if`. Ajoutons une autre ligne de sortie si la personne est assez âgée pour voter, demandant si elle s'est déjà inscrite:

````python
age = 19
if age >= 18:
  print("Vous êtes assez âgé pour voter !")
  print("Vous êtes-vous déjà inscrit sur les listes électorales ?")
````

Le test conditionnel réussit, et les deux appels `print()` sont indentés, donc les deux lignes sont imprimées:

```
Vous êtes assez âgé pour voter !
Vous êtes-vous déjà inscrit sur les listes électorales ?
```

#### Instructions `if-else`

Souvent, vous voudrez effectuer une action lorsqu'un test conditionnel réussit et une action différente dans tous les autres cas. La syntaxe `if-else` de Python rend cela possible. Un bloc `if-else` est similaire à une simple instruction `if`, mais l'instruction `else` vous permet de définir une action ou un ensemble d'actions qui sont exécutées lorsque le test conditionnel échoue.

Affichons le même message si la personne est assez âgée pour voter, mais ajoutons un message pour toute personne qui n'est pas assez âgée:

````python
age = 17
if age >= 18: # Le test échoue car 17 < 18
  print("Vous êtes assez âgé pour voter !")
  print("Vous êtes-vous déjà inscrit sur les listes électorales ?")
else: # Le bloc else est exécuté
  print("Désolé, vous êtes trop jeune pour voter.")
  print("Veuillez vous inscrire dès que vous aurez 18 ans !")
````

Si le test conditionnel réussit, le premier bloc d'appels `print()` indentés est exécuté. Si le test est évalué à `False`, le bloc `else` est exécuté. Comme l'âge est inférieur à 18 cette fois, le test conditionnel échoue et le code dans le bloc `else` est exécuté:

```
Désolé, vous êtes trop jeune pour voter.
Veuillez vous inscrire dès que vous aurez 18 ans !
```

Cette structure `if-else` fonctionne bien dans les situations où vous voulez que Python exécute toujours l'une des deux actions possibles.

#### La Chaîne `if-elif-else`

Souvent, vous devrez tester plus de deux situations possibles, et pour les évaluer, vous pouvez utiliser la syntaxe `if-elif-else` de Python. Python n'exécute qu'un seul bloc dans une chaîne `if-elif-else`. Il exécute chaque test conditionnel dans l'ordre jusqu'à ce qu'un réussisse. Lorsqu'un test réussit, le code suivant ce test est exécuté et Python saute le reste des tests.

De nombreuses situations réelles impliquent plus de deux conditions possibles. Par exemple, considérons une compagnie de bus qui applique des tarifs différents selon l'âge :
* Les enfants de moins de 6 ans voyagent gratuitement.
* Les personnes entre 6 et 18 ans paient 150 FCFA.
* Les personnes de 18 ans ou plus paient 300 FCFA.

Comment pouvons-nous utiliser une instruction `if` pour déterminer le tarif d'une personne ? Le code suivant teste la tranche d'âge d'une personne et imprime ensuite un message sur le prix du billet:

````python
age = 12
if age < 6: # Test si moins de 6 ans
  print("Votre ticket de bus est gratuit.")
elif age < 18: # Test si moins de 18 ans (et au moins 6 ans)
  print("Votre ticket de bus coûte 150 FCFA.")
else: # Pour tous les autres (18 ans et plus)
  print("Votre ticket de bus coûte 300 FCFA.")
````

Le test `if` vérifie si une personne a moins de 6 ans. Si le test réussit, un message approprié est imprimé et Python saute le reste des tests. La ligne `elif` est un autre test `if`, qui ne s'exécute que si le test précédent a échoué. À ce stade de la chaîne, nous savons que la personne a au moins 6 ans car le premier test a échoué. Si la personne a moins de 18 ans, un message approprié est imprimé et Python saute le bloc `else`. Si les tests `if` et `elif` échouent tous les deux, Python exécute le code dans le bloc `else`.

Dans cet exemple, le premier test (`age < 6`) est `False`. Le second test (`age < 18`) est `True` (12 est inférieur à 18), donc son code est exécuté. La sortie est une seule phrase, informant l'utilisateur du coût du billet:

```
Votre ticket de bus coûte 150 FCFA.
```

Plutôt que d'imprimer le prix dans le bloc `if-elif-else`, il serait plus concis de définir uniquement le prix à l'intérieur de la chaîne `if-elif-else` et d'avoir ensuite un simple appel `print()` qui s'exécute après l'évaluation de la chaîne:

````python
age = 12
if age < 6:
  prix = 0
elif age < 18:
  prix = 150
else:
  prix = 300
print(f"Votre ticket de bus coûte {prix} FCFA.")
````

Ce code produit la même sortie que l'exemple précédent, mais le but de la chaîne `if-elif-else` est plus restreint. Au lieu de déterminer un prix et d'afficher un message, il détermine simplement le prix du billet. En plus d'être plus efficace, ce code révisé est plus facile à modifier que l'approche originale.

#### Utilisation de Plusieurs Blocs `elif`

Vous pouvez utiliser autant de blocs `elif` que vous le souhaitez dans votre code. Par exemple, si la compagnie de bus mettait en œuvre une réduction pour les étudiants (disons 100 FCFA pour les moins de 25 ans avec une carte d'étudiant), vous pourriez ajouter un test conditionnel supplémentaire:

````python
age = 22
est_etudiant = True
if age < 6:
  prix = 0
elif age < 18:
  prix = 150
elif age < 25 and est_etudiant: # Test supplémentaire pour étudiant
  prix = 100
else: # Pour les 18+ non étudiants ou 25+
  prix = 300
print(f"Votre ticket de bus coûte {prix} FCFA.")
````

#### Omission du Bloc `else`

Python n'exige pas de bloc `else` à la fin d'une chaîne `if-elif`. Parfois, un bloc `else` est utile ; parfois, il est plus clair d'utiliser une instruction `elif` supplémentaire qui capture la condition spécifique d'intérêt:

````python
age = 70
if age < 6:
  prix = 0
elif age < 18:
  prix = 150
elif age < 65: # Tarif normal pour adultes < 65 ans
  prix = 300
elif age >= 65: # Tarif spécial pour seniors >= 65 ans
  prix = 200
print(f"Votre ticket de bus coûte {prix} FCFA.")
````

Le bloc `elif` supplémentaire à la fin assigne un prix de 200 FCFA lorsque la personne a 65 ans ou plus, ce qui est un peu plus clair que le bloc `else` général. Avec ce changement, chaque bloc de code doit passer un test spécifique pour être exécuté.

#### Tester Plusieurs Conditions

La chaîne `if-elif-else` est puissante, mais elle ne convient que lorsque vous avez besoin qu'un seul test réussisse. Dès que Python trouve un test qui réussit, il saute le reste des tests.

Cependant, il est parfois important de vérifier *toutes* les conditions d'intérêt. Dans ce cas, vous devez utiliser une série d'instructions `if` simples sans blocs `elif` ou `else`. Cette technique est logique lorsque plus d'une condition pourrait être `True`, et que vous voulez agir sur chaque condition qui est `True`.

Reconsidérons l'exemple du restaurant (pizzeria dans l'original). Si quelqu'un commande un plat avec plusieurs accompagnements, vous devrez vous assurer d'inclure tous les accompagnements. Adapté à un contexte camerounais, imaginons la commande d'un plat comme le "Poulet DG" avec des suppléments :

````python
supplements_commandes = ['plantain', 'riz', 'alloco'] # 'alloco' = bananes plantains frites
print("Préparation de votre Poulet DG avec :")
if 'plantain' in supplements_commandes:
  print("- Supplément Plantain")
if 'riz' in supplements_commandes:
  print("- Supplément Riz")
if 'alloco' in supplements_commandes:
  print("- Supplément Alloco")
print("\nPlat terminé !")
````

Nous commençons par une liste contenant les suppléments demandés. L'instruction `if` pour 'plantain' vérifie si le client a demandé du plantain. Si oui, un message est imprimé confirmant ce supplément. Le test pour 'riz' est une autre instruction `if` simple, pas une instruction `elif` ou `else`, donc ce test est exécuté que le test précédent ait réussi ou non. Le code pour 'alloco' vérifie si l'alloco a été demandé indépendamment des résultats des deux premiers tests. Ces trois tests indépendants sont exécutés à chaque fois que ce programme est lancé.

Comme chaque condition est évaluée, tous les suppléments demandés sont ajoutés:

```
Préparation de votre Poulet DG avec :
- Supplément Plantain
- Supplément Riz
- Supplément Alloco

Plat terminé !
```

Ce code ne fonctionnerait pas correctement si nous utilisions un bloc `if-elif-else`, car le code s'arrêterait après la réussite du premier test.

En résumé, si vous ne voulez exécuter qu'un seul bloc de code, utilisez une chaîne `if-elif-else`. Si plus d'un bloc de code doit être exécuté, utilisez une série d'instructions `if` indépendantes.

### Utilisation des Instructions `if` avec les Listes

Vous pouvez faire un travail intéressant lorsque vous combinez des listes et des instructions `if`. Vous pouvez surveiller les valeurs spéciales qui doivent être traitées différemment des autres valeurs de la liste. Vous pouvez gérer efficacement les conditions changeantes, comme la disponibilité de certains plats dans un restaurant tout au long de la journée.

#### Vérification des Éléments Spéciaux

Revenons à l'exemple du restaurant. Le restaurant affiche un message chaque fois qu'un ingrédient est ajouté à votre plat pendant sa préparation. Le code pour cette action peut être écrit très efficacement en faisant une liste des ingrédients demandés par le client et en utilisant une boucle pour annoncer chaque ingrédient au fur et à mesure de son ajout:

````python
ingredients_commandes = ['poisson_fume', 'gombo', 'huile_rouge'] # Ingrédients pour une sauce gombo
for ingredient in ingredients_commandes:
  print(f"Ajout de {ingredient}.")
print("\nSauce Gombo terminée !")
````

La sortie est simple car ce code est juste une simple boucle `for`:

```
Ajout de poisson_fume.
Ajout de gombo.
Ajout de huile_rouge.

Sauce Gombo terminée !
```

Mais que se passe-t-il si le restaurant est à court de gombo ? Une instruction `if` à l'intérieur de la boucle `for` peut gérer cette situation de manière appropriée:

````python
ingredients_commandes = ['poisson_fume', 'gombo', 'huile_rouge']
for ingredient in ingredients_commandes:
  if ingredient == 'gombo': # Vérifie si l'ingrédient est 'gombo'
    print("Désolé, nous n'avons plus de gombo pour le moment.")
  else: # Ajoute les autres ingrédients
    print(f"Ajout de {ingredient}.")
print("\nSauce Gombo terminée !")
````

Cette fois, nous vérifions chaque ingrédient demandé avant de l'ajouter. Le code avec `if` vérifie si la personne a demandé du gombo. Si oui, nous affichons un message les informant pourquoi ils ne peuvent pas avoir de gombo. Le bloc `else` garantit que tous les autres ingrédients seront ajoutés.

La sortie montre que chaque ingrédient demandé est traité de manière appropriée:

```
Ajout de poisson_fume.
Désolé, nous n'avons plus de gombo pour le moment.
Ajout de huile_rouge.

Sauce Gombo terminée !
```

#### Vérifier qu'une Liste n'est pas Vide

Nous avons fait une simple supposition sur chaque liste avec laquelle nous avons travaillé jusqu'à présent ; nous avons supposé que chaque liste contenait au moins un élément. Bientôt, nous laisserons les utilisateurs fournir les informations stockées dans une liste, nous ne pourrons donc pas supposer qu'une liste contient des éléments à chaque exécution d'une boucle. Dans cette situation, il est utile de vérifier si une liste est vide avant d'exécuter une boucle `for`.

Par exemple, vérifions si la liste des ingrédients commandés est vide avant de préparer le plat. Si la liste est vide, nous demanderons à l'utilisateur et nous assurerons qu'il souhaite un plat nature. Si la liste n'est pas vide, nous préparerons le plat comme dans les exemples précédents:

````python
ingredients_commandes = [] # Liste vide
if ingredients_commandes: # Vérifie si la liste contient des éléments
  for ingredient in ingredients_commandes:
    print(f"Ajout de {ingredient}.")
  print("\nPlat terminé !")
else: # Si la liste est vide
  print("Êtes-vous sûr de vouloir un plat sans ingrédients ?")
````

Cette fois, nous commençons avec une liste vide d'ingrédients commandés. Au lieu de sauter directement dans une boucle `for`, nous effectuons une vérification rapide. Lorsque le nom d'une liste est utilisé dans une instruction `if`, Python renvoie `True` si la liste contient au moins un élément ; une liste vide est évaluée à `False`. Si `ingredients_commandes` réussit le test conditionnel (c'est-à-dire n'est pas vide), nous exécutons la même boucle `for` que dans l'exemple précédent. Si le test conditionnel échoue (la liste est vide), nous imprimons un message demandant au client s'il veut vraiment un plat nature.

La liste est vide dans ce cas, donc la sortie demande si l'utilisateur veut vraiment un plat nature:

```
Êtes-vous sûr de vouloir un plat sans ingrédients ?
```

#### Utilisation de Plusieurs Listes

Les clients demanderont à peu près n'importe quoi, surtout quand il s'agit de nourriture. Et si un client voulait des frites dans son Ndolé ? Vous pouvez utiliser des listes et des instructions `if` pour vous assurer que votre entrée a du sens avant d'agir dessus.

Surveillons les demandes d'ingrédients inhabituelles avant de préparer un plat. L'exemple suivant définit deux listes. La première est une liste d'ingrédients disponibles au restaurant, et la seconde est la liste des ingrédients que le client a demandés. Cette fois, chaque élément dans `ingredients_commandes` est vérifié par rapport à la liste des ingrédients disponibles avant d'être ajouté au plat:

````python
ingredients_disponibles = ['feuilles_ameres', 'arachides', 'crevettes', 'poisson_fume', 'huile_rouge']
ingredients_commandes = ['feuilles_ameres', 'frites', 'crevettes'] # Demande inhabituelle : frites

for ingredient in ingredients_commandes: # Parcourt les ingrédients commandés
  if ingredient in ingredients_disponibles: # Vérifie si l'ingrédient est disponible
    print(f"Ajout de {ingredient}.")
  else: # Si l'ingrédient n'est pas disponible
    print(f"Désolé, nous n'avons pas de {ingredient}.")

print("\nPlat terminé !")
````

Nous définissons d'abord une liste d'ingrédients disponibles. Ensuite, nous créons une liste d'ingrédients qu'un client a demandés, y compris la demande inhabituelle de 'frites'. Nous parcourons ensuite la liste des ingrédients demandés. À l'intérieur de la boucle, nous vérifions d'abord si chaque ingrédient demandé est réellement dans la liste des ingrédients disponibles. Si c'est le cas, nous ajoutons cet ingrédient. Si l'ingrédient demandé n'est pas dans la liste des ingrédients disponibles, le bloc `else` s'exécutera. Le bloc `else` imprime un message informant l'utilisateur quels ingrédients ne sont pas disponibles.

Ce code produit une sortie claire et informative:

```
Ajout de feuilles_ameres.
Désolé, nous n'avons pas de frites.
Ajout de crevettes.

Plat terminé !
```
